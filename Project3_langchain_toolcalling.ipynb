**Project 3 : Building a LangChain Calculator with Google Gemini Flash**
<br>Code tried by me </br>

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import initialize_agent, Tool, AgentType
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain.agents import initialize_agent, Tool, AgentType


# Importing required Python modules
import re

def calculator_tool(func):
    """
    A decorator that could be used to enhance the calculator function with logging, validation, etc.
    """
    def wrapper(expression: str):
        print(f"Processing expression: {expression}")

        # Call the original function (the calculator)
        if not re.match(r'^[\d+\-*/().\s]+$', expression):  # Ensure safe characters
            return "Invalid input: Only numbers and operators are allowed."

        result = func(expression)

        print(f"Result: {result}")

        return result

    return wrapper

@calculator_tool
def calculator(expression: str) -> str:
    """
    Placeholder function that simulates calling the Google Gemini Flash model.
    This would normally interact with Gemini Flash to compute the result.
    """
    return str(eval(expression))

calculator_tool = Tool(
    name="Calculator",
    func=calculator,
    description="Use this tool to calculate any mathematical expression provided in text."
)

llm=ChatGoogleGenerativeAI(model='gemini-1.5-flash',verbose=True)

tools = [calculator_tool]
agent = initialize_agent(tools, llm, agent_type=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True)
# llm_with_tools = llm.bind_tools([calculator])

# Example input text: Calculate "2 + 2" or "Product of 6 and 8"
input_text = input("Input numbers to calculate : ")

# Run the agent to compute the result
response = agent.run(input_text)

print("Response:", response)
```

```python
from langchain_core.tools import tool

@tool
def multiply(a: int, b: int) -> int:
    """Multiply a and b.

    Args:
        a: first int
        b: second int
    """
    return a * b

llm_with_tools = llm.bind_tools([multiply])
```

```python
from langchain.tools import Tool

# Example tool for addition
def add_numbers(a, b):
    return a + b

math_tool = Tool(
    name="AdditionTool",
    func=lambda x: f"The result is: {add_numbers(*map(int, x.split()))}",
    description="Adds two numbers. Input should be two integers separated by a space."
)
```

```python
from langchain.agents import initialize_agent, AgentType

# List of tools
tools = [math_tool]

# Initialize the agent
agent = initialize_agent(
    tools=tools,                        # Provide the tools
    llm=llm,                            # LLM for fallback
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,  # Agent type
    verbose=True                        # Enable debugging output
)
```

```python
# Query the agent
response1 = agent.run("What is 10 + 20?")  # Uses the tool
print(response1)
```
